```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def show(img, title=''):
    # OpenCV is BGR; matplotlib expects RGB
    if img.ndim == 3:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        plt.imshow(img)
    else:
        plt.imshow(img, cmap='gray', vmin=0, vmax=255)
    plt.title(title); plt.axis('off'); plt.show()
```

```python
img=cv2.imread('//content/imggg.webp')
show(img, 'Original')
print("Shape: ", img.shape)
print("Data type: ", img.dtype)
print("Min pixel value: ", img.min())
print("Max pixel value: ",img.max())
cv2.imwrite('blueobjectcopy.jpg', img)
```

```python
b,g,r= cv2.split(img)
show(b, 'Blue')
show(g, 'Green')
show(r, 'Red')
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
lower = np.array([100, 60, 30])
upper = np.array([130, 255, 255])
mask  = cv2.inRange(hsv, lower, upper)
show(mask, 'object mask')
result = cv2.bitwise_and(img, img, mask=mask)
show(result, 'Isolated Object')
result = cv2.bitwise_and(img, img, mask=mask)
show(result, 'Isolated Object')

```
